## Resultat

Analysen jämför patienternas mätvärden före och efter medicinering.

För glukos undersöks skillnaden mellan det genomsnittliga värdet före och efter medicinering.

För puls undersöks på samma sätt förändringen mellan perioden före och efter medicinering.

Graferna visar förändringen över tid och markerar när medicineringen började.

```python
selected_patient_id = "P001"
```

```python
selected_glucose = results[selected_patient_id]
```

```python
selected_pulse = pulse_results["P003"]
```

```python
print(f"Patient: P003")
print(f"Puls före medicinering: {selected_pulse['unmedicated_avg']:.0f} bpm")
print(f"Puls efter medicinering: {selected_pulse['medicated_avg']:.0f} bpm")
print(f"Skillnad: {selected_pulse['difference']:.0f} bpm")
print(f"Förändring: {selected_pulse['change']}")
```

```python
print(f"Patient: P001")
print(f"Glukos före medicinering: {selected_glucose['unmedicated_avg']:.1f} mmol/L")
print(f"Glukos efter medicinering: {selected_glucose['medicated_avg']:.1f} mmol/L")
print(f"Skillnad: {selected_glucose['difference']:.1f} mmol/L")
print(f"Förändring: {selected_glucose['change']}")
```
